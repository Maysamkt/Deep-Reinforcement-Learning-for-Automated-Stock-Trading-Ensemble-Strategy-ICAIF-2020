### Passo 1: Verificar aceleração de hardware (GPU)

```python
import subprocess

print("=== Verificação de Hardware (GPU) ===")
try:
    res = subprocess.run(["nvidia-smi", "--query-gpu=name,memory.total", "--format=csv,noheader"], capture_output=True, text=True)
    if res.returncode == 0 and res.stdout.strip():
        print(f"GPU Detectada: {res.stdout.strip()}")
    else:
        print("AVISO: GPU não detectada via nvidia-smi. O treinamento rodará em CPU (mais lento).")
except Exception as e:
    print(f"nvidia-smi não disponível: {e}")
```
